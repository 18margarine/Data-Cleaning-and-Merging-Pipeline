```python
import pandas as pd
import os
```

### **Constants**

```python
INPUT_DIR = '../data'
OUTPUT_DIR = '../cleaned_data'

ACTIVITY_TYPE_MAP = {
    'Play': 'Playing',
    'Walk': 'Walking',
    'Rest': 'Resting'
}
```

```python
pet_activity_df = pd.read_csv(f'{INPUT_DIR}/pet_activities.csv')
pet_health_df = pd.read_csv(f'{INPUT_DIR}/pet_health.csv')
users_df = pd.read_csv(f'{INPUT_DIR}/users.csv')
```

### **Cleaning Pet Activity data**

```python
pet_activity_df['date'] = pd.to_datetime(pet_activity_df['date'])
pet_activity_df['duration_minutes'] = pet_activity_df['duration_minutes'].replace('-', 0).astype(int)
```

Normalise activity type labels

```python
pet_activity_df['activity_type'] = pet_activity_df['activity_type'].replace(ACTIVITY_TYPE_MAP)
```

### **Cleaning Pet Health data**

```python
pet_health_df['visit_date'] = pd.to_datetime(pet_health_df['visit_date'])
pet_health_df.rename(columns={'visit_date': 'date'}, inplace=True)
```

Adding columns for a consistent schema

```python
pet_health_df['activity_type'] = 'Health'
pet_health_df['duration_minutes'] = 0
```

### **Combine and merge**

```python
pet_combine_df = pd.concat([pet_activity_df, pet_health_df], ignore_index=True)
all_df = pet_combine_df.merge(users_df, on='pet_id', how='left')
```

### **Generating the clean data**

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)
output_path = f'{OUTPUT_DIR}/clean_data.csv'
all_df.to_csv(output_path, index=False)

print(f'Saved {len(all_df):,} rows to {output_path}')
```
